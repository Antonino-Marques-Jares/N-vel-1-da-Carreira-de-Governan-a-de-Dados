```python
# ============================================================
# GERA 03_politica_acesso_qualidade.xlsx FORMATADO
# ============================================================
!pip install openpyxl --quiet

import pandas as pd
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter

# ---------- DADOS ----------

# Aba 1: Política
politica = [
    ["elemento", "conteudo"],
    ["Propósito", "Estabelecer regras claras, mensuráveis e auditáveis para o acesso e a qualidade dos dados de clientes do SwiftBank, garantindo conformidade com a LGPD, consistência entre áreas e uso seguro e ético dos dados."],
    ["Escopo – Cobre", "Sistemas: CRM, core bancário, data lake (bronze/silver/gold), data warehouse. Áreas: Marketing, Crédito, Financeiro, Operações, Comercial, Jurídico/Compliance. Tipos de dados: cadastrais, transacionais, de crédito, de consentimento e de uso."],
    ["Escopo – NÃO cobre", "Dados de colaboradores (RH), dados de fornecedores, dados públicos de mercado, dados anonimizados de forma irreversível."],
    ["CDO", "Aprova a política e responde pela governança de dados perante o board."],
    ["DGO (Data Governance Office)", "Mantém a política, monitora indicadores e coordena exceções."],
    ["Data Owners por domínio", "Aprovam acessos e respondem pela qualidade no domínio (Comercial, Crédito, Financeiro, Marketing, Operações, Jurídico)."],
    ["Data Stewards (Business e Technical)", "Definem regras de negócio e implementam controles técnicos."],
    ["Data Custodians", "Armazenam, mantêm e garantem disponibilidade das bases."],
    ["Usuários", "Seguem a política, usam dados apenas para a finalidade aprovada e reportam problemas."],
    ["Data Champions", "Disseminam a cultura de dados e boas práticas nas áreas."],
    ["D1 – Aprovação de acesso", "Todo acesso a dados pessoais deve ser aprovado pelo Data Owner do domínio antes da liberação. | Responsável: Data Owner | Frequência: Por solicitação | Registro: Ticket no Service Desk + e-mail formal | Consequência: Suspensão do acesso e advertência formal."],
    ["D2 – Criptografia", "Campos com dados pessoais devem ser criptografados em repouso e em trânsito. | Responsável: Segurança da Informação | Frequência: Contínua | Registro: Inventário de segurança | Consequência: Bloqueio do campo e abertura de incidente."],
    ["D3 – Regras de qualidade antes do go-live", "Toda base deve ter regras de qualidade documentadas no catálogo antes de ir para produção. | Responsável: Business Steward + Technical Steward | Frequência: Antes do go-live | Registro: Catálogo de dados | Consequência: Base não entra em produção."],
    ["D4 – Validação do consentimento", "O campo consentimento_marketing deve ser validado mensalmente quanto à consistência. | Responsável: Business Steward Marketing | Frequência: Mensal | Registro: Relatório de qualidade | Consequência: Correção obrigatória em 5 dias úteis."],
    ["D5 – Data Literacy", "Usuários devem concluir o treinamento de Data Literacy antes de acessar dados pessoais. | Responsável: DGO + Data Champions | Frequência: Anual | Registro: Plataforma de treinamento | Consequência: Acesso bloqueado até conclusão."],
    ["D6 – Exceções formais", "Exceções à política devem ser solicitadas formalmente e aprovadas pelo DGO com aval do Data Owner. | Responsável: DGO + Data Owner | Frequência: Por solicitação | Registro: Formulário de exceção | Consequência: Exceção negada automaticamente se não formalizada."],
    ["Base legal", "LGPD (Lei 13.709/2018), Marco Civil da Internet, regulações do BACEN."],
    ["Como solicitar exceção", "Preencher formulário no Service Desk com: justificativa, prazo, área envolvida, dados afetados."],
    ["Quem aprova exceção", "Data Owner do domínio (aval técnico) + DGO (aval de governança)."],
    ["Prazo de validade da exceção", "Máximo de 90 dias, renovável mediante nova justificativa."],
    ["Registro de exceções", "Toda exceção é registrada no catálogo de exceções e revisada trimestralmente."],
]

# Aba 2: Plano de Implantação
plano = [
    ["passo", "acao_concreta", "responsavel", "cronograma", "integracao_com_metas"],
    ["1. Comunicação direcionada", "Workshop com Marketing antes do go-live da nova regra de Cliente Ativo", "DGO + Business Steward Marketing", "Semana 1–2", "Meta de 100% da área treinada"],
    ["1. Comunicação direcionada", "Apresentação da política ao comitê de Crédito", "CDO + Data Owner Crédito", "Semana 2", "Aprovação formal em ata"],
    ["1. Comunicação direcionada", "E-mail institucional com resumo da política para todas as áreas", "DGO", "Semana 1", "Confirmação de leitura"],
    ["2. Responsáveis e cronograma", "Nomeação formal dos Data Owners e Stewards por domínio", "CDO", "Semana 1–3", "Publicação em organograma"],
    ["2. Responsáveis e cronograma", "Reunião quinzenal com Data Champions", "DGO", "Quinzenal", "Ata e plano de ação"],
    ["2. Responsáveis e cronograma", "Definição de SLA de correção de qualidade (5 dias úteis)", "DGO + Data Owners", "Semana 3", "Inclusão no catálogo"],
    ["3. Integração com metas e rituais", "Indicador de qualidade entra no OKR trimestral da Diretoria de Crédito", "CDO + Diretoria Crédito", "Trimestre 1", "OKR formalizado"],
    ["3. Integração com metas e rituais", "Revisão trimestral da política no comitê de governança", "DGO", "Trimestral", "Ata de reunião"],
    ["3. Integração com metas e rituais", "Auditoria interna anual de conformidade", "Jurídico + DGO", "Anual", "Relatório de auditoria"],
]

# Aba 3: Indicadores
indicadores = [
    ["#", "indicador", "formula", "fonte_do_dado", "meta", "responsavel"],
    [1, "% de dados mestres atualizados no prazo", "(clientes atualizados no prazo / total) × 100", "Catálogo + CRM", "≥ 95%", "Data Owner Cadastro"],
    [2, "Número de inconsistências por mês", "Contagem de registros com erro de qualidade", "Relatório de qualidade", "≤ 10/mês", "Business Steward"],
    [3, "% de áreas que cumprem revisão trimestral", "(áreas que revisaram / total) × 100", "Atas do comitê", "100%", "DGO"],
    [4, "Tempo médio de correção de inconsistência", "Soma dos tempos / nº de correções", "Service Desk", "≤ 5 dias úteis", "Data Owner"],
    [5, "Engajamento em treinamentos de Data Literacy", "(treinados / total) × 100", "Plataforma de treinamento", "≥ 90%", "DGO + Data Champions"],
    [6, "% de bases com Data Owner definido", "(bases com owner / total) × 100", "Catálogo de dados", "100%", "DGO"],
    [7, "Número de exceções ativas", "Contagem de exceções em vigor", "Catálogo de exceções", "≤ 5 simultâneas", "DGO"],
]

# Aba 4: Cenários
cenarios = [
    ["cenario", "descricao", "politica_aplicavel", "falhas_sem_politica", "papeis_envolvidos", "acoes_corretivas", "diretriz_que_previne"],
    ["1. Compartilhamento com consultoria de cobrança", "Crédito quer compartilhar base de inadimplentes com consultoria externa.", "D1, D2, D6", "Vazamento de dados; compartilhamento sem base legal; multa LGPD.", "Data Owner Crédito, DPO, Jurídico, Segurança", "Aprovação formal + contrato com cláusula LGPD + anonimização parcial.", "D1 e D6"],
    ["2. Dúvida sobre consentimento_marketing", "Marketing quer usar dados sem consentimento; Crédito discorda.", "D4, D5, D1", "Uso indevido; multa LGPD; dano reputacional.", "Business Steward Marketing, Data Owner Marketing, DPO, Jurídico", "Bloqueio do uso até parecer jurídico; revisão da base legal.", "D4"],
    ["3. Extração emergencial de empréstimos", "Diretoria pede extração urgente para reunião em 2 horas.", "D1, D6, D2", "Acesso não autorizado; vazamento; perda de rastreabilidade.", "Data Owner Crédito, DGO, Segurança", "Aprovação verbal registrada + formalização em 24h + log de auditoria.", "D1 e D6"],
    ["4. Clientes duplicados no CRM", "Engenharia detecta 15% de clientes duplicados.", "D3, D4", "Relatórios incorretos; decisões erradas; retrabalho.", "Business Steward Crédito, Technical Steward, Data Owner", "Deduplicação com regra documentada; atualização do catálogo.", "D3"],
]

# ---------- CRIAÇÃO DO ARQUIVO ----------

wb = Workbook()

# Estilos
header_font = Font(bold=True, color="FFFFFF", size=12)
header_fill = PatternFill("solid", fgColor="1F4E78")
sub_fill = PatternFill("solid", fgColor="D9E1F2")
border = Border(
    left=Side(style="thin"), right=Side(style="thin"),
    top=Side(style="thin"), bottom=Side(style="thin")
)
wrap = Alignment(wrap_text=True, vertical="top")

def escreve_aba(ws, dados, titulo):
    ws.title = titulo
    for i, linha in enumerate(dados, start=1):
        for j, valor in enumerate(linha, start=1):
            cell = ws.cell(row=i, column=j, value=valor)
            cell.border = border
            cell.alignment = wrap
            if i == 1:
                cell.font = header_font
                cell.fill = header_fill
            elif j == 1:
                cell.font = Font(bold=True)
                cell.fill = sub_fill
    # Ajustar largura das colunas
    for col in ws.columns:
        max_len = max(len(str(c.value)) if c.value else 0 for c in col)
        ws.column_dimensions[get_column_letter(col[0].column)].width = min(max_len + 2, 60)

# Criar abas
escreve_aba(wb.active, politica, "Política")
escreve_aba(wb.create_sheet(), plano, "Plano_Implantacao")
escreve_aba(wb.create_sheet(), indicadores, "Indicadores")
escreve_aba(wb.create_sheet(), cenarios, "Cenarios_Diagnostico")

# Salvar
wb.save("03_politica_acesso_qualidade.xlsx")
print("✅ Arquivo '03_politica_acesso_qualidade.xlsx' gerado com sucesso!")

# Download no Colab
try:
    from google.colab import files
    files.download("03_politica_acesso_qualidade.xlsx")
    print("📥 Download iniciado.")
except ImportError:
    print("📁 Arquivo salvo na pasta atual.")
```
